```python
import matplotlib.pyplot as plt

# Data
quant = ["FP16", "q8_0", "q6_k", "q5_k_m", "q4_k_m"]
accuracy = [100, 95, 85, 75, 65]
tokens_sec = [0.8, 1.5, 2.5, 3.5, 5.0]
ram = [10, 7, 5, 4, 2.8]

plt.figure(figsize=(10, 6))

# Color palette (visually distinct, inspired by file_context_0 style)
accuracy_color = "#d7263d"
tokens_color = "#1f77b4"

# Plot accuracy (primary y-axis)
ax1 = plt.gca()
ax1.set_xlabel("Quantization Level", fontsize=12)
ax1.set_ylabel("Relative Accuracy (%)", fontsize=12, color=accuracy_color)
l1 = ax1.plot(
    quant,
    accuracy,
    marker="o",
    linewidth=2,
    markersize=8,
    color=accuracy_color,
    label="Accuracy",
)
ax1.tick_params(axis="y", labelcolor=accuracy_color)
for i, val in enumerate(accuracy):
    ax1.text(
        quant[i],
        val + 1,
        f"{val}%",
        ha="center",
        va="bottom",
        fontsize=10,
        color=accuracy_color,
    )

# Plot tokens/sec (secondary y-axis)
ax2 = ax1.twinx()
ax2.set_ylabel("Tokens per Second", fontsize=12, color=tokens_color)
l2 = ax2.plot(
    quant,
    tokens_sec,
    marker="s",
    linestyle="--",
    linewidth=2,
    markersize=8,
    color=tokens_color,
    label="Tokens/sec",
)
ax2.tick_params(axis="y", labelcolor=tokens_color)
for i, val in enumerate(tokens_sec):
    ax2.text(
        quant[i],
        val + 0.12,
        f"{val}",
        ha="center",
        va="bottom",
        fontsize=10,
        color=tokens_color,
    )

plt.title(
    "Accuracy–Throughput Trade-off Across Quantization Levels", fontsize=14, pad=15
)
ax1.grid(alpha=0.3)
plt.xticks(fontsize=11)
ax1.yticks = plt.yticks(fontsize=11)
ax2.yticks = ax2.set_yticklabels(ax2.get_yticks(), fontsize=11)

# Legend, tight layout
lines = l1 + l2
labels = [l.get_label() for l in lines]
plt.legend(lines, labels, loc="upper center", fontsize=10, frameon=False)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
quant = ["FP16", "q8_0", "q6_k", "q5_k_m", "q4_k_m"]
accuracy = [100, 95, 85, 75, 65]
tokens_sec = [0.8, 1.5, 2.5, 3.5, 5.0]

# Figure
plt.figure(figsize=(10, 6))

# Scatter with visually distinct color and outline
plt.scatter(
    tokens_sec,
    accuracy,
    color="#d7263d",
    edgecolor="#1f77b4",
    s=120,
    linewidth=2,
    zorder=3,
    label="Quantization Setting",
)

# Line to highlight the trend ("Pareto frontier" visual impact)
plt.plot(
    tokens_sec,
    accuracy,
    color="#1f77b4",
    linestyle="--",
    linewidth=2,
    zorder=2,
    alpha=0.3,
)

# Annotate each point with the quantization method
for i, label in enumerate(quant):
    plt.text(
        tokens_sec[i],
        accuracy[i] + 1.2,
        label,
        ha="center",
        fontsize=11,
        color="black",
        weight="bold",
        zorder=4,
    )

# Labels and title
plt.xlabel("Tokens per Second", fontsize=12)
plt.ylabel("Relative Accuracy (%)", fontsize=12)
plt.title("Pareto Frontier: Accuracy vs Inference Speed", fontsize=14, pad=15)

# Grid and ticks
plt.grid(alpha=0.3, zorder=0)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Data
hardware = ["Arduino", "Raspberry Pi", "Laptop", "Workstation"]
quant = ["FP16", "q8_0", "q6_k", "q5_k_m", "q4_k_m"]

feasibility = np.array(
    [
        [0, 0, 0, 0, 0],  # Arduino
        [0, 0, 1, 1, 1],  # Raspberry Pi
        [0, 1, 2, 2, 2],  # Laptop
        [2, 2, 2, 2, 2],  # Workstation
    ]
)

# Convert to DataFrame for seaborn
df = pd.DataFrame(feasibility, index=hardware, columns=quant)

# Plot
plt.figure(figsize=(8, 5))
sns.heatmap(
    df,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={"label": "Feasibility Level"},
)

plt.xlabel("Quantization Method")
plt.ylabel("Target Hardware")
plt.title("Hardware Feasibility Across Quantization Levels")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Data
batch_size = np.array([1, 2, 4, 8])
quant_factor = np.array([1.0, 0.8, 0.6, 0.5, 0.4])  # FP16 → q4

X, Y = np.meshgrid(batch_size, quant_factor)
latency = (1200 / Y) * (X**0.5)

# Figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# Surface plot
surface = ax.plot_surface(
    X,
    Y,
    latency,
    cmap="viridis",
    edgecolor="none",
    antialiased=True,
    alpha=0.95,
)

# Color bar
cbar = fig.colorbar(surface, shrink=0.6, aspect=12)
cbar.set_label("Latency (ms)")

# Labels and title
ax.set_xlabel("Batch Size", labelpad=10)
ax.set_ylabel("Quantization Factor\n(1.0 = FP16, 0.4 = q4)", labelpad=12)
ax.set_zlabel("Inference Latency (ms)", labelpad=10)

ax.set_title(
    "Estimated CPU Inference Latency Surface\nAcross Batch Size and Quantization Levels",
    pad=20,
)

# View angle (critical for readability)
ax.view_init(elev=28, azim=235)

plt.tight_layout()
plt.show()
```
